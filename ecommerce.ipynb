```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from imblearn.over_sampling import SMOTE

# 2. Separate Features (X) and Target (y)
X = df.drop('PurchaseStatus', axis=1)
y = df['PurchaseStatus']

# 3. Define Column Categories
categorical_cols = ['Gender', 'ProductCategory', 'PreferredDevice', 
                    'Region', 'ReferralSource', 'CustomerSegment']

numerical_cols = ['Age', 'AnnualIncome', 'NumberOfPurchases', 'TimeSpentOnWebsite', 
                  'CustomerTenureYears', 'LastPurchaseDaysAgo', 'DiscountsAvailed', 
                  'SessionCount', 'CustomerSatisfaction']

# Note: 'LoyaltyProgram' is already binary (0/1), so we will bypass it during scaling/encoding.

# 4. Build the Preprocessing Transformer
preprocessor = ColumnTransformer(
    transformers=[
        # Scale continuous variables (Critical for distance-based kNN)
        ('num', StandardScaler(), numerical_cols),
        
        # Encode categoricals, dropping the first column to prevent multicollinearity
        ('cat', OneHotEncoder(drop='first'), categorical_cols)
    ],
    remainder='passthrough' # Leaves the LoyaltyProgram column as is
)

# Apply the transformations to the feature set
X_processed = preprocessor.fit_transform(X)

# 5. Train-Test Split (Stratified)
# Stratify ensures the original imbalanced ratio is preserved in both sets before resampling
X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.3, random_state=42, stratify=y
)

# We ONLY apply SMOTE to the training data to prevent data leakage
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# 7. Verification Output
print(f"Original Training Target Shape: {y_train.value_counts().to_dict()}")
print(f"Resampled Training Target Shape: {y_train_resampled.value_counts().to_dict()}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import f1_score, roc_auc_score
import pandas as pd

# 1. Initialize the Models
# We set random_state for reproducibility and increase max_iter for Logistic Regression to ensure convergence.
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "k-Nearest Neighbors (kNN)": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Naive Bayes": GaussianNB()
}

# Note: Gaussian Naive Bayes requires dense arrays. 
# If our OneHotEncoder created a sparse matrix, we convert it back here.
if hasattr(X_train_resampled, "toarray"):
    X_train_resampled = X_train_resampled.toarray()
    X_test = X_test.toarray()

# 2. Create a list to store the evaluation metrics
results = []

# 3. The Training and Evaluation Loop
for name, model in models.items():
    # Fit the model on the SMOTE-balanced training data
    model.fit(X_train_resampled, y_train_resampled)
    
    # Predict the discrete classes (for F1-Score)
    y_pred = model.predict(X_test)
    
    # Predict the probabilities for the positive class (for ROC AUC)
    # predict_proba returns an array of shape (n_samples, n_classes), so we slice [:, 1] for class '1'
    y_proba = model.predict_proba(X_test)[:, 1] 
    
    # Calculate the metrics against the unseen test data
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_proba)
    
    # Append to our results list
    results.append({
        "Model": name,
        "F1-Score": round(f1, 4),
        "ROC AUC": round(roc_auc, 4)
    })

# 4. Display the Results
# Convert the results dictionary into a Pandas DataFrame for clean tabular viewing
results_df = pd.DataFrame(results).sort_values(by="ROC AUC", ascending=False)

print("--- Model Evaluation Results ---")
print(results_df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_curve, ConfusionMatrixDisplay,classification_report

# Set general plot style for professional aesthetics
sns.set_theme(style="whitegrid")

# ---------------------------------------------------------
# Plot 1: Combined ROC Curve
# ---------------------------------------------------------
plt.figure(figsize=(10, 8))

for name, model in models.items():
    # Get probability of the positive class (Purchase = 1)
    y_proba = model.predict_proba(X_test)[:, 1]
    
    # Calculate False Positive Rate (fpr) and True Positive Rate (tpr)
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    
    # Fetch the AUC score from our results dataframe to include in the legend
    auc_score = results_df.loc[results_df['Model'] == name, 'ROC AUC'].values[0]
    
    # Plot the curve
    plt.plot(fpr, tpr, lw=2, label=f'{name} (AUC = {auc_score:.2f})')

# Plot the baseline (random guessing)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Guess')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=12, fontweight='bold')
plt.ylabel('True Positive Rate', fontsize=12, fontweight='bold')
plt.title('Receiver Operating Characteristic (ROC) Curve', fontsize=14, fontweight='bold')
plt.legend(loc="lower right")
plt.show()

# ---------------------------------------------------------
# Plot 2: Model Performance Bar Chart
# ---------------------------------------------------------
# Melt the DataFrame so Seaborn can easily plot side-by-side bars
results_melted = results_df.melt(id_vars="Model", var_name="Metric", value_name="Score")

plt.figure(figsize=(10, 6))
sns.barplot(x="Model", y="Score", hue="Metric", data=results_melted, palette="viridis")

plt.ylim(0, 1.1)
plt.title("Model Performance Comparison (F1-Score vs ROC AUC)", fontsize=14, fontweight='bold')
plt.ylabel("Score", fontsize=12, fontweight='bold')
plt.xlabel("Algorithm", fontsize=12, fontweight='bold')
plt.legend(loc='lower right')
plt.show()

# ---------------------------------------------------------
# Plot 3: Confusion Matrices Grid
# ---------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for ax, (name, model) in zip(axes, models.items()):
    # Predict discrete classes
    y_pred = model.predict(X_test)
    
    # Generate and plot confusion matrix
    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred, 
        ax=ax, 
        cmap='Blues', 
        colorbar=False,
        display_labels=['No Purchase', 'Purchase']
    )
    ax.set_title(f'{name}', fontsize=12, fontweight='bold')
    ax.grid(False) # Turn off the seaborn gridlines inside the matrix

plt.suptitle("Confusion Matrices by Model", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.subplots_adjust(top=0.92) # Adjust spacing for the main title
plt.show()

print('\nclassificattion report:')
print(classification_report(y_test,y_pred))
```

```python
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler

# 2. Isolate Continuous Variables (The Random Vectors)
# We will focus on three strictly continuous behavioral metrics
features = ['TimeSpentOnWebsite', 'AnnualIncome', 'SessionCount']
X_raw = df[features].dropna() 

# 3. Standardize the Vectors
# Crucial step: Income is in thousands, Time is in minutes. 
# Scaling ensures the covariance matrices don't become singular or numerically unstable during MLE.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# 4. Formulate and Fit the Gaussian Mixture Model
# We will hypothesize k=3 latent sub-populations for this initial test.
# covariance_type='full' means each component has its own general covariance matrix.
k = 3
gmm = GaussianMixture(n_components=k, covariance_type='full', random_state=42)

# The Expectation-Maximization (EM) algorithm executes here
gmm.fit(X_scaled)

# 5. Extract the Maximum Likelihood Estimates
weights = gmm.weights_       
means = gmm.means_           
covariances = gmm.covariances_ 

# 6. Output the Theoretical Parameters
print("--- EM Algorithm Convergence ---")
print(f"Converged: {gmm.converged_}")
print(f"Number of Iterations to Converge: {gmm.n_iter_}\n")

print("--- Estimated Parameters (Scaled) ---")
for i in range(k):
    print(f"Latent Class {i+1}:")
    print(f"  Mixing Weight (pi): {weights[i]:.4f}")
    print(f"  Mean Vector (mu): {means[i]}")
    print(f"  Covariance Matrix (Sigma):\n{covariances[i]}\n")
```
